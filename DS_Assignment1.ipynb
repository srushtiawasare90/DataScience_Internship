```python
df['quantity'] = df['quantity'].fillna(df['quantity'].median())

df['price'] = df['price'].fillna(df['price'].median())
```

```python
df['category'] = df['category'].fillna('Unknown')

df['payment_method'] = df['payment_method'].fillna(
    df['payment_method'].mode()[0]
)
```

```python
total_revenue = df['Total Amount'].sum()
```

```python
df['revenue'] = df['price'] * df['quantity'] * (1 - df['discount'])
```

```python
# Total Revenue
total_revenue = df['revenue'].sum()

# Total Orders
total_orders = df['order_id'].nunique()

# Average Order Value
avg_order_value = total_revenue / total_orders

# Units Sold
units_sold = df['quantity'].sum()

# Print Results
print("Total Revenue:", total_revenue)
print("Total Orders:", total_orders)
print("Average Order Value:", avg_order_value)
print("Units Sold:", units_sold)
```

```python
top_category = (
    df.groupby('category')['revenue']
      .sum()
      .idxmax()
)
```

```python
top_product = (
    df.groupby('product_id')['revenue']
      .sum()
      .idxmax()
)
```

```python
kpi = pd.DataFrame({
    'KPI': [
        'Revenue',
        'Orders',
        'Average Order Value',
        'Units Sold',
        'Top Category',
        'Top Product'
    ],
    'Value': [
        total_revenue,
        total_orders,
        avg_order_value,
        units_sold,
        top_category,
        top_product
    ]
})

print(kpi)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(14, 8))

# KPI Cards
axes[0,0].axis('off')

kpi_text = f"""
Total Revenue: {total_revenue:,.2f}

Total Orders: {total_orders}

Average Order Value: {avg_order_value:,.2f}

Units Sold: {units_sold}

Top Category: {top_category}
"""

axes[0,0].text(
    0.05,
    0.5,
    kpi_text,
    fontsize=12,
    verticalalignment='center'
)

axes[0,0].set_title("KPI Summary")

# Revenue by Category
category_sales = (
    df.groupby('category')['revenue']
      .sum()
      .sort_values(ascending=False)
)

category_sales.plot(
    kind='bar',
    ax=axes[0,1]
)

axes[0,1].set_title("Revenue by Category")
axes[0,1].set_xlabel("Category")
axes[0,1].set_ylabel("Revenue")

# Monthly Sales Trend
monthly_sales.plot(
    kind='line',
    marker='o',
    ax=axes[1,0]
)

axes[1,0].set_title("Monthly Sales Trend")
axes[1,0].set_xlabel("Month")
axes[1,0].set_ylabel("Revenue")

# Payment Method Distribution
payment = df['payment_method'].value_counts()

payment.plot(
    kind='pie',
    autopct='%1.1f%%',
    ax=axes[1,1]
)

axes[1,1].set_title("Payment Method Distribution")
axes[1,1].set_ylabel("")

plt.tight_layout()

plt.savefig(
    r"C:\Users\Admin\OneDrive\DataScience_Internship\DataScience_Week1\outputs\dashboard\dashboard_mockup.png"
)

plt.show()
```
